## Relativistic Doppler Equation

The relativistic Doppler equation describes the relationship between the redshift of light and the velocity of a celestial object moving away from the observer.

The equation is:

$$
1+z=\sqrt{\frac{1+\beta}{1-\beta}}
$$

Where:

- $z$ represents redshift.
- $\beta = \frac{v}{c}$ represents the velocity as a fraction of the speed of light.
- $v$ is the velocity of the celestial object.
- $c$ is the speed of light.

```python
import sympy as sp

z = sp.symbols('z')
```

```python
beta = sp.symbols('beta')

equation = sp.Eq(1 + z, sp.sqrt((1 + beta)/(1 - beta)))

equation
```

```python
solution = sp.solve(equation, beta)

solution
```

```python
beta_formula = sp.simplify(solution[0])

beta_formula
```

```python
derivative = sp.diff(beta_formula, z)

derivative
```

```python
simplified_derivative = sp.simplify(derivative)

simplified_derivative
```

## Mathematical Interpretation

The derivative of β with respect to redshift (z) is positive for non-negative redshift values.

This indicates that as redshift increases, the velocity of a celestial object also increases.

The velocity approaches the speed of light but never reaches or exceeds it for finite positive redshift values. This supports the principle of Einstein's Special Theory of Relativity.


## Data Acquisition

We obtained the Stellar Classification Dataset – SDSS17 from Kaggle. The dataset contains astronomical observations and information used for classifying celestial objects such as stars, galaxies and quasars.

The original dataset is stored in the `data/raw` folder of our project.

```python
import pandas as pd
import numpy as np
import os

project_path = '..'
raw_path = '../data/raw'
df = pd.read_csv(os.path.join(raw_path, 'star_classification.csv'))
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
required_columns = ['u', 'g', 'r', 'i', 'z', 'redshift', 'class']

df = df[required_columns]

df.head()
```

```python
df = df.dropna()

print("Dataset Shape After Removing Missing Values:", df.shape)
```

```python
light_columns = ['u', 'g', 'r', 'i', 'z']

df = df[(df[light_columns] > 0).all(axis=1)]

print("Dataset Shape After Removing Invalid Values:", df.shape)
```

```python
negative_redshift = (df['redshift'] < 0).sum()

print("Number of Negative Redshift Values:", negative_redshift)
```

## Handling Negative Redshift Values

Negative redshift values are retained in our dataset because they represent blueshift, where celestial objects move towards the observer.

The relativistic Doppler equation supports negative values of redshift and gives negative velocity fractions for these objects.

```python
processed_path = os.path.join(project_path, 'data', 'processed')

output_path = os.path.join(processed_path, 'sdss_clean.csv')

df.columns = df.columns.str.lower()

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
```
